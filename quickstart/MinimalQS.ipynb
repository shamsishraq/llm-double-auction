```python
#1 — Imports
import requests
import random
import re
import time
import copy
import os
import json
from pathlib import Path
from datetime import datetime
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#2 — Ollama connection and model
OLLAMA_URL = "http://localhost:11434/api/generate"
MODEL = "phi3:mini"

def ask_phi(prompt, temperature=0.7):
    payload = {
        "model": MODEL,
        "prompt": prompt,
        "stream": False,
        "options": {"temperature": temperature, "num_predict": 30}
    }
    r = requests.post(OLLAMA_URL, json=payload, timeout=120)
    r.raise_for_status()
    return r.json()["response"].strip()

test_response = ask_phi("Reply with exactly the number 5.", temperature=0)
print(test_response)
```

```python
#3 — Experiment parameters
TREATMENT = "Minimal"
N_BUYERS = 11
N_SELLERS = 11
ROUNDS = 5
MIN_VALUE = 0.75
MAX_VALUE = 3.25
STEP = 0.25
EQUILIBRIUM_PRICE = 2.00
EQUILIBRIUM_QUANTITY = 6
MAX_DECISIONS_PER_PERIOD = 60
STAGNATION_LIMIT = 15
TEMPERATURE = 0.7
SCHEDULER_SEED = 42
VALUATION_SEED = 42
FINAL_CALLS = True
TEST_RUN = False

if TEST_RUN:
    ROUNDS = 1
    MAX_DECISIONS_PER_PERIOD = 2
    STAGNATION_LIMIT = 2
    FINAL_CALLS = False

# Create a unique output directory for this run before the experiment starts.
REPO_ROOT = Path.cwd().resolve()
for candidate in [REPO_ROOT, *REPO_ROOT.parents]:
    if (candidate / "results").exists() and (candidate / "quickstart").exists():
        REPO_ROOT = candidate
        break
RESULTS_ROOT = REPO_ROOT / "results" / "replications" / TREATMENT
EXPERIMENT_ID = datetime.now().strftime("%Y%m%d_%H%M%S")
OUTPUT_DIR = RESULTS_ROOT / EXPERIMENT_ID
OUTPUT_DIR.mkdir(parents=True, exist_ok=False)

print("Treatment:", TREATMENT)
print("Experiment ID:", EXPERIMENT_ID)
print("Output directory:", OUTPUT_DIR)
```

```python
#4 — Reservation values and costs
buyer_values = [MAX_VALUE - i * STEP for i in range(N_BUYERS)]
seller_costs = [MIN_VALUE + i * STEP for i in range(N_SELLERS)]
print("Buyer values:", buyer_values)
print("Seller costs:", seller_costs)
```

```python
#5 — Randomly assign private values/costs to trader IDs
valuation_rng = random.Random(VALUATION_SEED)
buyer_assignments = buyer_values.copy()
seller_assignments = seller_costs.copy()
valuation_rng.shuffle(buyer_assignments)
valuation_rng.shuffle(seller_assignments)
agents = {}
for i in range(N_BUYERS):
    agent_id = f"B{i+1}"
    agents[agent_id] = {"id": agent_id, "role": "buyer", "reservation": buyer_assignments[i]}
for i in range(N_SELLERS):
    agent_id = f"S{i+1}"
    agents[agent_id] = {"id": agent_id, "role": "seller", "reservation": seller_assignments[i]}
print("\nTRADER ASSIGNMENTS")
print(pd.DataFrame(agents.values()).to_string(index=False))
```

```python
#6 — Persistent private trader memory
private_memory = {}
for agent_id in agents:
    private_memory[agent_id] = {
        "periods": [],
        "current_actions": [],
        "current_trades": [],
        "current_profit": 0.0,
        "total_profit": 0.0
    }
```

```python
#7 — Public cross-period history
public_history = []
```

```python
#8 — Create a fresh trading-period state
def new_period_state():
    return {
        "active": set(agents.keys()),
        "current_bids": {},
        "current_asks": {},
        "transactions": [],
        "decision_log": [],
        "market_tape": [],
        "decision_count": 0,
        "stagnation": 0,
        "period": None
    }
```

```python
#9 — Compact public market state
def public_market_state(state):
    bids = sorted(state["current_bids"].items(), key=lambda x: (-x[1], x[0]))
    asks = sorted(state["current_asks"].items(), key=lambda x: (x[1], x[0]))
    current_prices = [round(t["Price"], 4) for t in state["transactions"]]
    return {
        "standing_bids": bids,
        "standing_asks": asks,
        "current_prices": current_prices,
        "recent_tape": state["market_tape"][-20:],
        "previous_periods": public_history[-4:]
    }
```

```python
#10 — Compact private trader memory
def private_memory_state(agent_id):
    memory = private_memory[agent_id]
    period_summary = []
    for p in memory["periods"][-4:]:
        period_summary.append({
            "Period": p["Period"],
            "Actions": p["Actions"][-6:],
            "Trades": p["Trades"],
            "Profit": p["Profit"]
        })
    return {
        "previous_periods": period_summary,
        "current_actions": memory["current_actions"][-6:],
        "current_trades": memory["current_trades"],
        "total_profit": round(memory["total_profit"], 4)
    }
```

```python
#11 — Prompt builder: MINIMAL
PROMPT_MODE = "minimal"

def build_prompt(agent_id, state):
    agent = agents[agent_id]
    public = public_market_state(state)
    private = private_memory_state(agent_id)
    if agent["role"] == "buyer":
        role_text = f"""
You are buyer {agent_id}.
Your private maximum willingness to pay is {agent["reservation"]:.2f}.
You may purchase at most one unit in this trading period.
If you buy at price P, your profit is {agent["reservation"]:.2f} - P.
You cannot buy above your maximum willingness to pay.
"""
        action_text = """
You may submit a BID at or below your maximum willingness to pay.
You may ACCEPT the lowest standing ask if it is acceptable to you.
You may PASS.
"""
        format_text = "BID X, ACCEPT, or PASS"
    else:
        role_text = f"""
You are seller {agent_id}.
Your private minimum willingness to accept is {agent["reservation"]:.2f}.
You may sell at most one unit in this trading period.
If you sell at price P, your profit is P - {agent["reservation"]:.2f}.
You cannot sell below your minimum willingness to accept.
"""
        action_text = """
You may submit an ASK at or above your minimum willingness to accept.
You may ACCEPT the highest standing bid if it is acceptable to you.
You may PASS.
"""
        format_text = "ASK X, ACCEPT, or PASS"
    prompt = f"""
{role_text}
Other traders' private information is unknown to you.
Use the public market information available to you.
{action_text}
PUBLIC MARKET INFORMATION:
Standing bids: {public["standing_bids"]}
Standing asks: {public["standing_asks"]}
Current transaction prices: {public["current_prices"]}
Recent public market tape: {public["recent_tape"]}
Previous-period public summaries: {public["previous_periods"]}
YOUR PRIVATE MEMORY:
{private}
Respond with exactly one action:
{format_text}
Do not provide an explanation.
"""
    return prompt.strip()
```

```python
#12 — Parse LLM action
def parse_action(response):
    text = response.strip().upper()
    if "ACCEPT" in text:
        return {"action": "ACCEPT", "price": None}
    bid_match = re.search(r"BID\s+([0-9]+(?:\.[0-9]+)?)", text)
    if bid_match:
        return {"action": "BID", "price": float(bid_match.group(1))}
    ask_match = re.search(r"ASK\s+([0-9]+(?:\.[0-9]+)?)", text)
    if ask_match:
        return {"action": "ASK", "price": float(ask_match.group(1))}
    if "PASS" in text:
        return {"action": "PASS", "price": None}
    return {"action": "INVALID", "price": None}
```

```python
#13 — Add public market event
def add_market_event(state, event):
    state["market_tape"].append(event)
```

```python
#14 — Find best standing ask
def best_ask(state):
    if not state["current_asks"]:
        return None
    return min(state["current_asks"].items(), key=lambda x: (x[1], x[0]))
```

```python
#15 — Find best standing bid
def best_bid(state):
    if not state["current_bids"]:
        return None
    return max(state["current_bids"].items(), key=lambda x: (x[1], x[0]))
```

```python
#16 — Experiment event log helper
experiment_log = []

def log_event(event_type, period, decision=None, agent_id=None, role=None, **fields):
    row = {
        "Event Type": event_type,
        "Treatment": TREATMENT,
        "Experiment ID": EXPERIMENT_ID,
        "Timestamp": datetime.now().isoformat(timespec="seconds"),
        "Period": period,
        "Decision": decision,
        "Agent": agent_id,
        "Role": role
    }
    row.update(fields)
    experiment_log.append(row)
    return row

def json_text(value):
    return json.dumps(value, default=list, ensure_ascii=False)
```

```python
#17 — Execute a trade
def execute_trade(state, buyer_id, seller_id, price, initiating_agent, trade_type):
    buyer_value = agents[buyer_id]["reservation"]
    seller_cost = agents[seller_id]["reservation"]
    buyer_profit = buyer_value - price
    seller_profit = price - seller_cost
    trade = {
        "Period": state["period"], "Buyer": buyer_id, "Seller": seller_id,
        "Price": round(price, 4), "Buyer Value": buyer_value,
        "Seller Cost": seller_cost, "Buyer Profit": round(buyer_profit, 4),
        "Seller Profit": round(seller_profit, 4),
        "Initiating Agent": initiating_agent, "Trade Type": trade_type
    }
    state["transactions"].append(trade)
    state["active"].discard(buyer_id)
    state["active"].discard(seller_id)
    state["current_bids"].pop(buyer_id, None)
    state["current_asks"].pop(seller_id, None)
    private_memory[buyer_id]["current_trades"].append(trade)
    private_memory[seller_id]["current_trades"].append(trade)
    private_memory[buyer_id]["current_profit"] += buyer_profit
    private_memory[seller_id]["current_profit"] += seller_profit
    add_market_event(state, f"TRADE: {buyer_id}-{seller_id} @ {price:.2f}")
    log_event(
        "TRADE", state["period"], state["decision_count"], initiating_agent,
        agents[initiating_agent]["role"], Buyer=buyer_id, Seller=seller_id,
        Price=round(price, 4), Buyer_Value=buyer_value, Seller_Cost=seller_cost,
        Buyer_Profit=round(buyer_profit, 4), Seller_Profit=round(seller_profit, 4),
        Trade_Type=trade_type
    )
    return trade
```

```python
#18 — Process one trader decision
def process_agent(agent_id, state):
    if agent_id not in state["active"]:
        return False
    agent = agents[agent_id]
    prompt = build_prompt(agent_id, state)
    public_before = public_market_state(state)
    private_before = private_memory_state(agent_id)
    response = ask_phi(prompt, temperature=TEMPERATURE)
    parsed = parse_action(response)
    state["decision_count"] += 1
    log_entry = {
        "Period": state["period"], "Decision": state["decision_count"],
        "Agent": agent_id, "Role": agent["role"], "Response": response,
        "Action": parsed["action"], "Price": parsed["price"]
    }
    state["decision_log"].append(log_entry)
    private_memory[agent_id]["current_actions"].append(log_entry)
    log_event(
        "DECISION", state["period"], state["decision_count"], agent_id, agent["role"],
        Reservation=agent["reservation"], Action=parsed["action"],
        Quoted_Price=parsed["price"], Response=response, Prompt=prompt,
        Public_Market=json_text(public_before), Private_Memory=json_text(private_before)
    )
    if parsed["action"] == "PASS":
        add_market_event(state, f"{agent_id} passes")
        return False
    if parsed["action"] == "INVALID":
        add_market_event(state, f"{agent_id} gives invalid response")
        return False
    if agent["role"] == "buyer":
        if parsed["action"] == "BID":
            bid = parsed["price"]
            if bid > agent["reservation"]:
                add_market_event(state, f"{agent_id} invalid BID {bid:.2f}")
                return False
            old_bid = state["current_bids"].get(agent_id)
            if old_bid == bid:
                add_market_event(state, f"{agent_id} repeats BID {bid:.2f}")
                return False
            state["current_bids"][agent_id] = bid
            if old_bid is None:
                add_market_event(state, f"{agent_id} submits BID {bid:.2f}")
            else:
                add_market_event(state, f"{agent_id} revises BID {old_bid:.2f} -> {bid:.2f}")
            ask = best_ask(state)
            if ask is not None and bid >= ask[1]:
                seller_id, ask_price = ask
                execute_trade(state, agent_id, seller_id, ask_price, agent_id, "Bid Crossed Ask")
            return True
        if parsed["action"] == "ACCEPT":
            ask = best_ask(state)
            if ask is not None and ask[1] <= agent["reservation"]:
                seller_id, ask_price = ask
                execute_trade(state, agent_id, seller_id, ask_price, agent_id, "Buyer Accepted Ask")
                return True
            add_market_event(state, f"{agent_id} attempted ACCEPT with no acceptable ask")
            return False
    if agent["role"] == "seller":
        if parsed["action"] == "ASK":
            ask = parsed["price"]
            if ask < agent["reservation"]:
                add_market_event(state, f"{agent_id} invalid ASK {ask:.2f}")
                return False
            old_ask = state["current_asks"].get(agent_id)
            if old_ask == ask:
                add_market_event(state, f"{agent_id} repeats ASK {ask:.2f}")
                return False
            state["current_asks"][agent_id] = ask
            if old_ask is None:
                add_market_event(state, f"{agent_id} submits ASK {ask:.2f}")
            else:
                add_market_event(state, f"{agent_id} revises ASK {old_ask:.2f} -> {ask:.2f}")
            bid = best_bid(state)
            if bid is not None and bid[1] >= ask:
                buyer_id, bid_price = bid
                execute_trade(state, buyer_id, agent_id, bid_price, agent_id, "Ask Crossed Bid")
            return True
        if parsed["action"] == "ACCEPT":
            bid = best_bid(state)
            if bid is not None and bid[1] >= agent["reservation"]:
                buyer_id, bid_price = bid
                execute_trade(state, buyer_id, agent_id, bid_price, agent_id, "Seller Accepted Bid")
                return True
            add_market_event(state, f"{agent_id} attempted ACCEPT with no acceptable bid")
            return False
    return False
```

```python
#19 — Create a randomized round-robin decision order
def decision_order(active_agents, rng):
    order = list(active_agents)
    rng.shuffle(order)
    return order
```

```python
#20 — Run one trading period
def run_period(period_number):
    state = new_period_state()
    state["period"] = period_number
    rng = random.Random(SCHEDULER_SEED + period_number)
    print(f"\n{'='*70}\nPERIOD {period_number}\n{'='*70}")
    while len(state["active"]) >= 2 and state["decision_count"] < MAX_DECISIONS_PER_PERIOD and state["stagnation"] < STAGNATION_LIMIT:
        order = decision_order(state["active"], rng)
        for agent_id in order:
            if len(state["active"]) < 2 or state["decision_count"] >= MAX_DECISIONS_PER_PERIOD:
                break
            changed = process_agent(agent_id, state)
            state["stagnation"] = 0 if changed else state["stagnation"] + 1
            if state["stagnation"] >= STAGNATION_LIMIT:
                break
    if state["decision_count"] >= MAX_DECISIONS_PER_PERIOD:
        close_reason = "decision limit reached"
    elif state["stagnation"] >= STAGNATION_LIMIT:
        close_reason = "stagnation limit reached"
    else:
        close_reason = "fewer than two active traders"
    print(f"Main market closed after {state['decision_count']} decisions ({close_reason}).")
    if FINAL_CALLS and len(state["active"]) >= 2:
        print("\nFINAL CALLS")
        final_order = decision_order(state["active"], rng)
        for agent_id in final_order:
            if agent_id not in state["active"] or len(state["active"]) < 2:
                continue
            process_agent(agent_id, state)
    print("\nTRADES")
    for trade in state["transactions"]:
        print(f"{trade['Buyer']}-{trade['Seller']} @ {trade['Price']:.2f} ({trade['Trade Type']})")
    if not state["transactions"]:
        print("No trades.")
    return state
```

```python
#21 — Summarize public period information
def period_public_summary(state):
    prices = [t["Price"] for t in state["transactions"]]
    return {
        "Period": state["period"], "Trades": len(state["transactions"]),
        "Transaction Prices": [round(p, 4) for p in prices],
        "Average Price": round(np.mean(prices), 4) if prices else None,
        "Price Range": (round(min(prices), 4), round(max(prices), 4)) if prices else None,
        "Decisions": state["decision_count"]
    }
```

```python
#22 — Save private period memory
def save_private_period_memory(state):
    for agent_id in agents:
        memory = private_memory[agent_id]
        memory["periods"].append({
            "Period": state["period"],
            "Actions": memory["current_actions"].copy(),
            "Trades": memory["current_trades"].copy(),
            "Profit": round(memory["current_profit"], 4)
        })
        memory["total_profit"] += memory["current_profit"]
        memory["current_actions"] = []
        memory["current_trades"] = []
        memory["current_profit"] = 0.0
```

```python
#23 — Test initial market state
test_state = new_period_state()
print("Active traders:", len(test_state["active"]))
print("Standing bids:", test_state["current_bids"])
print("Standing asks:", test_state["current_asks"])
print("Transactions:", test_state["transactions"])
```

```python
#24 — Run the five-period experiment
all_states = []
all_decision_logs = []
all_transactions = []
experiment_start = time.time()
for period in range(1, ROUNDS + 1):
    state = run_period(period)
    all_states.append(state)
    all_decision_logs.extend(state["decision_log"])
    all_transactions.extend(state["transactions"])
    summary = period_public_summary(state)
    public_history.append(summary)
    save_private_period_memory(state)
experiment_time = time.time() - experiment_start
transactions_df = pd.DataFrame(all_transactions)
decisions_df = pd.DataFrame(all_decision_logs)
print(f"\nTotal runtime: {experiment_time/60:.2f} minutes")
print(f"Total decisions: {len(decisions_df)}")
print(f"Total trades: {len(transactions_df)}")
```

```python
#25 — Maximum theoretical surplus
def max_surplus():
    buyers = sorted([a["reservation"] for a in agents.values() if a["role"] == "buyer"], reverse=True)
    sellers = sorted([a["reservation"] for a in agents.values() if a["role"] == "seller"])
    return sum(buyer - seller for buyer, seller in zip(buyers, sellers) if buyer > seller)
MAX_SURPLUS = max_surplus()
print("Maximum positive-surplus allocation:", MAX_SURPLUS)
```

```python
#26 — Smith alpha
def smith_alpha(prices, equilibrium_price=EQUILIBRIUM_PRICE):
    if len(prices) == 0:
        return np.nan
    prices = np.array(prices, dtype=float)
    return 100 * np.sqrt(np.mean((prices - equilibrium_price) ** 2)) / equilibrium_price
```

```python
#27 — Period diagnostics
period_results = []
for state in all_states:
    prices = [t["Price"] for t in state["transactions"]]
    actual_surplus = sum(t["Buyer Profit"] + t["Seller Profit"] for t in state["transactions"])
    row = {
        "Period": state["period"], "Trades": len(state["transactions"]),
        "Average Price": round(np.mean(prices), 4) if prices else np.nan,
        "Alpha": round(smith_alpha(prices), 4) if prices else np.nan,
        "Surplus": round(actual_surplus, 4),
        "Efficiency (%)": round(100 * actual_surplus / MAX_SURPLUS, 4) if MAX_SURPLUS > 0 else np.nan,
        "Decisions": state["decision_count"], "Stagnation": state["stagnation"]
    }
    period_results.append(row)
    log_event("PERIOD_SUMMARY", state["period"], None, None, None, **row)
period_results_df = pd.DataFrame(period_results)
print("\nPERIOD RESULTS")
print(period_results_df.to_string(index=False))
```

```python
#28 — Transaction diagnostics and overall summary
if not transactions_df.empty:
    transactions_df["Total Surplus"] = transactions_df["Buyer Profit"] + transactions_df["Seller Profit"]
    transactions_df["Positive Surplus"] = transactions_df["Buyer Value"] > transactions_df["Seller Cost"]
    print("\nTRANSACTION DIAGNOSTICS")
    print(transactions_df[["Period", "Buyer", "Seller", "Price", "Buyer Value", "Seller Cost", "Buyer Profit", "Seller Profit", "Total Surplus", "Positive Surplus", "Trade Type"]].to_string(index=False))
overall_summary = {
    "Total Trades": len(transactions_df),
    "Total Decisions": len(decisions_df),
    "Average Transaction Price": transactions_df["Price"].mean() if not transactions_df.empty else np.nan,
    "Overall Surplus": transactions_df["Total Surplus"].sum() if not transactions_df.empty else 0,
    "Overall Efficiency (%)": 100 * transactions_df["Total Surplus"].sum() / (MAX_SURPLUS * ROUNDS) if not transactions_df.empty else 0,
    "Runtime (minutes)": experiment_time / 60
}
log_event("EXPERIMENT_SUMMARY", "ALL", None, None, None, **overall_summary)
print("\nOVERALL SUMMARY")
for key, value in overall_summary.items():
    print(f"{key}: {value:.4f}" if isinstance(value, (float, np.floating)) else f"{key}: {value}")
```
